```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np


df = pd.read_csv(
    r"C:\Users\DELL\Downloads\task1_marketplace_delivery_sla_raw(in).csv"
)
df
```

```python
print("Shape:", df.shape)
print(df.info())
print(df.isnull().sum())
```

```python
df.columns = df.columns.str.strip()
df
```

```python
date_columns = [
    "order_date",
    "shipped_at",
    "delivered_at",
    "updated_at"
]

for col in date_columns:
    df[col] = pd.to_datetime(
        df[col],
        format='mixed',
        errors="ignore",
        dayfirst=True
    )
df
```

```python
df.info()
```

```python
text_columns = [
    "courier",
    "city",
    "payment_status",
    "order_status",
    "is_test_order"
]

for col in text_columns:
    df[col] = df[col].str.strip().str.lower()

df
```

```python
df["promised_days"] = pd.to_numeric(
    df["promised_days"],
    errors="coerce"
)
df
```

```python
df = df.sort_values("updated_at")

df = df.drop_duplicates(
    subset="order_id",
    keep="last")
df
```

```python
df.shape
```

```python
df["courier"] = df["courier"].replace({
    "shipquick": "ShipQuick",
    "fastbee": "FastBee",
    "bluedart": "BlueDart",
    "blue dart": "BlueDart"
})
df
```

```python
print(df.isnull().sum())
```

```python
df.dropna(inplace=True)
df.shape
```

```python
df['actual_transit_days'] = (
    df["delivered_at"] -
    df["shipped_at"]
).dt.days
df
```

```python
data = df.copy()
data['promised_days'] = data['promised_days'].abs()  #since promised days can't be negative
data
```

```python

data["delivery_status"] = np.where(
    data["actual_transit_days"] <= data["promised_days"],
    "On Time",
    "Late"
)
data
```

```python
may_orders = data[
    (df["delivered_at"] >= "2026-05-01") &
    (df["delivered_at"] < "2026-06-01")
].copy()

may_orders = may_orders[may_orders["is_test_order"] == "no"]

may_orders = may_orders[
    may_orders["order_status"] == "delivered"
]
```

```python
may_orders
```

```python
courier_summary = (
    may_orders.groupby("courier")
    .agg(
        eligible_orders=("order_id", "count"),
        on_time_orders=("delivery_status", lambda x: (x == "On Time").sum())
    )
    .reset_index()
)
print(courier_summary,'\n')

courier_summary["on_time_rate"] = (
    courier_summary["on_time_orders"] / courier_summary["eligible_orders"]
)

print(courier_summary)
```

```python
plt.figure(figsize=(5, 4))

sns.barplot(
    data=courier_summary,
    x="courier",
    y="on_time_rate"
)
plt.title("On-Time Delivery Rate by Courier")
plt.xlabel("Courier Partner")
plt.ylabel("On-Time Delivery Rate")
plt.show()
```

```python
courier_summary.plot(
    x="courier",
    y=["eligible_orders", "on_time_orders"],
    kind="bar",
    figsize=(8, 5)
)
plt.title("Eligible Orders vs On-Time Orders")
plt.xlabel("Courier")
plt.ylabel("Number of Orders")
plt.show()
```
